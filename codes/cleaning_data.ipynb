```python
import os
import glob
import pandas as pd
```

```python
#input dir w/ comment files
cwd = os.getcwd()
comment_dir = 'raw_comments'
comment_out_dir = 'cropped_comments'

#date range
end_date_raw = '2024-10-6'
end_date = pd.to_datetime(end_date_raw)

#grabbing files
raw_comment_files = glob.glob(os.path.join(cwd, comment_dir, '*.csv'))

aggregate_raw_length = 0
aggregate_filtered_length = 0
cleaned_comments = []
#loop thorugh each of the files and pull the data
for f in raw_comment_files:
    #read in csv
    raw_df = pd.read_csv(f)
    length = len(raw_df['published_at'].values)
    aggregate_raw_length += length

    #grabbing the date published and updated
    raw_df['published_at'] = pd.to_datetime(raw_df['published_at'])
    raw_df['updated_at'] = pd.to_datetime(raw_df['updated_at'])

    #if the comment was published or made after the cutoff time, remove it
    mask = (raw_df['published_at'] < end_date_raw) & (raw_df['updated_at'] < end_date_raw) & (raw_df['text'].str.len() > 0)
    # date_mask = 
    filtered_df = raw_df[mask]
    filtered_length = len(filtered_df['published_at'].values)
    aggregate_filtered_length += filtered_length
    print(f"Filename: {f}")
    print(f"\t{length - filtered_length} rows have been removed due to being published or updated after {end_date_raw} or due to empty text.")

    base, ext = os.path.splitext(f)
    parts = base.split('\\')
    filtered_file_name = parts[-1] + '_filtered' + ext
    filtered_path = os.path.join(cwd, comment_out_dir, filtered_file_name)
    print(f"Printing filtered data to {filtered_path}")

    filtered_df.to_csv(filtered_path)

print(f"The total length of the raw data was {aggregate_raw_length}.\nThere are {aggregate_filtered_length} filtered comments remaining.")
    
```
